```python
import pandas as pd
import numpy as np 
  
# column name for runs  
run_col = "fold_id" 

summary_rows = []
missing = []

# iterate over all datasets and settings 
  
for setting in SETTINGS.keys():
    for index, dataset in  enumerate(small_datasets): 
      print(f"dataset is {dataset}")  
      setting_dir = LOGS_ROOT / setting
      
      # If the directory is not present, then continue 
      if not setting_dir.exists():
          missing.append(f"[Missing setting folder] {setting_dir}")
          continue 
      
      TARGET_LR = lr
      no_of_client = parameters_dict[setting]["total_center"][index]
      
      no_of_fold = total_fold[index] 
      TARGET_BS =  parameters_dict[setting]["total_batch"][index] 
      
      print(f"The number of client is {no_of_client}")
      print(f"The batch sie is {TARGET_BS}") 
      
      dataset_dir = setting_dir/dataset   
      results_path = dataset_dir / input_log_file
      
      # If the log file for a particular dataset and setting is not present, then continue 
      if not results_path.exists():
          missing.append(f"[Missing results.csv] {results_path}")
          continue
      
      # read the log file from the specified directory 
      df = pd.read_csv(results_path)
      needed_cols = {"seed", "lr", "batch_size", "c_idx", "brier", run_col}
      
      # If the required columns are missing, then store the dataset name and continue 
      missing_cols = needed_cols - set(df.columns)
      if missing_cols:
          missing.append(f"[Missing columns {missing_cols}] {results_path}")
          continue
        
      # transform the pandas string columns to numeric 
      df["seed"] = pd.to_numeric(df["seed"], errors="coerce")
      df["lr"] = pd.to_numeric(df["lr"], errors="coerce")
      df["batch_size"] = pd.to_numeric(df["batch_size"], errors="coerce")
      df["c_idx"] = pd.to_numeric(df["c_idx"], errors="coerce")
      df["brier"] = pd.to_numeric(df["brier"], errors="coerce")
      df[run_col] = pd.to_numeric(df[run_col], errors="coerce")
      
      RUN_IDS = list(range(no_of_fold))
      filtered = df[
          (df["seed"] == TARGET_SEED) &
          (np.isclose(df["lr"], TARGET_LR)) &
          (df["batch_size"] == TARGET_BS) &
          (df[run_col].isin(RUN_IDS))
      ].copy() 
      
      # we have to ensure fold_id 0..4 are present
      filtered = filtered.sort_values(run_col)
      filtered.columns = filtered.columns.astype(str).str.strip()
      
      # If filtering does not provide any rows, we will log the dataset name in missing array and continue 
      if filtered.empty:
            print(df.head(2)) 
            missing.append(
                f"[No rows after filter] {results_path} "
                f"(seed={TARGET_SEED}, lr={TARGET_LR}, bs={TARGET_BS}, runs={sorted(RUN_IDS)})"
            )
            continue
      
      # we have to iterate as there are three models FedPDNN, FedPLSTM, and FedPAttn 
      for model_name in sorted(filtered["model"].dropna().unique()):
          sub = filtered[filtered["model"] == model_name].copy() 

          # If more than required rows, we will take last 3 rows
          if len(sub) > no_of_fold:
              sub = sub.tail(no_of_fold)

          # If less than required rows, we will log the dataset name in missing array and continue 
          elif len(sub) < no_of_fold:
              missing.append(
                  f"[Expected {no_of_client} rows, got {len(sub)}] {results_path} "
                  f"(seed={TARGET_SEED}, lr={TARGET_LR}, bs={TARGET_BS}, runs={sorted(RUN_IDS)})"
              )
              continue 
            
          # computation of mean and standard deviation of c-index
          c_mean = sub["c_idx"].mean()
          c_std  = sub["c_idx"].std(ddof=1)  
          
          # computation of mean and standard deviation of brier score
          b_mean = sub["brier"].mean()
          b_std  = sub["brier"].std(ddof=1)
         
          # computation of mean and standard deviation of nbll score
          nbll_mean = sub["nbll"].mean()
          nbll_std  = sub["nbll"].std(ddof=1)
          
          # computation of mean and standard deviation of D-calibration score
          p_value_mean = sub["p_value"].mean()
          p_value_std  = sub["p_value"].std(ddof=1)
           
          # computation of mean and standard deviation of MAE_PO score
          mae_po_mean = sub["weighted_MAE_PO"].mean()
          mae_po_std  = sub["weighted_MAE_PO"].std(ddof=1)
          
          # computation of mean and standard deviation of experiments time 
          time_mean = sub["time_counter"].mean()
          time_std  = sub["time_counter"].std(ddof=1)
          
          # computation of mean and standard deviation of memory usage
          column = "memory" if "memory" in sub.columns else "memory "
          memory_mean = sub[column].mean()
          memory_std  = sub[column].std(ddof=1) 

          # append c-index, brier score, time and memory usage in array for future use
          summary_rows.append({
                "setting": setting,
                "dataset": dataset,
                "model": model_name,
                "seed": TARGET_SEED,
                "lr": TARGET_LR,
                "batch_size": TARGET_BS,
                "n_runs": len(sub),

                # optional: nice formatted columns
                "cindex_mean_std": f"{c_mean:.4f} ± {c_std:.4f}",
                "brier_mean_std":  f"{b_mean:.4f} ± {b_std:.4f}",
                "nbll_mean_std":  f"{nbll_mean:.4f} ± {nbll_std:.4f}",
                "p_value_mean_std":  f"{p_value_mean:.4f} ± {p_value_std:.4f}",
                "mae_po_mean_std":  f"{mae_po_mean:.4f} ± {mae_po_std:.4f}",
                "time_mean_std":  f"{time_mean:.4f} ± {time_std:.4f}",
                "memory_mean_std":  f"{memory_mean:.4f} ± {memory_std:.4f}",
          })

# Printing issues in the experimetns if there is any 
if missing:
    print("\n⚠️ Issues found:")
    for m in missing[:50]:
        print(" -", m)
        
    if len(missing) > 50:
        print(f" ... and {len(missing)-50} more") 

# Experiments summary stored in CSV file
if summary_rows: 
    summary_df = pd.DataFrame(summary_rows).sort_values(["setting", "dataset"])
    out_path = LOGS_ROOT / destination_file_path
    summary_df.to_csv(out_path, index=False)

    print(f"Saved: {out_path}")
    print(f"Rows in summary: {len(summary_df)}")
```
